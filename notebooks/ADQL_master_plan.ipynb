# Selecting a Region

```python
cone_query = """SELECT
TOP 10 source_id
FROM gaiadr2.gaia_source
WHERE 1=CONTAINS(
POINT(ra, dec),
CIRCLE(88.8, 7.4, 0.08333))"""
```

```python
cone_job = Gaia.launch_job(cone_query)
cone_job
```

```python
cone_results = cone_job.get_results()
cone_results
```

# Exercise
When you are debugging queries like this, you can use TOP to limit the size of the results, but then you still don’t know how big the results will be.

An alternative is to use COUNT, which asks for the number of rows that would be selected, but it does not return them.

In the previous query, replace TOP 10 source_id with COUNT(source_id) and run the query again. How many stars has Gaia identified in the cone we searched?

```python
count_cone_query = """SELECT
COUNT(source_id)
FROM gaiadr2.gaia_source
WHERE 1=CONTAINS(
POINT(ra, dec),
CIRCLE(88.8, 7.4, 0.08333))"""

count_cone_job = Gaia.launch_job(count_cone_query)
count_cone_results = count_cone_job.get_results()
count_cone_results
```

# Join

```python
result = Gaia.launch_job('SELECT TOP 10 * from gaiadr2.vari_classifier_result')
result.get_data()
```

gaiadr2.vari_classifier_result has best_class_name and best_class_score. But the best_class_names are just an abbreviation. Let's look at what they mean 

```python
result = Gaia.launch_job('SELECT DISTINCT class_name, class_description from gaiadr2.vari_classifier_class_definition')
result.get_results()
```

```python
result = Gaia.launch_job('SELECT TOP 10 * from gaiadr2.gaia_source')
result.get_data()
```

```python
result = Gaia.launch_job("""SELECT source.source_id, ra, dec, pmra, pmdec, best_class_name, best_class_score 
FROM gaiadr2.gaia_source as source
JOIN gaiadr2.vari_classifier_result as var
ON source.SOURCE_ID = var.SOURCE_ID
WHERE var.best_class_name = 'CEP' AND 
var.best_class_score>0.98""").get_results()
```
